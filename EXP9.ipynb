```python
# EXPERIMENT 9
# Remove Punctuations from a String

import string

input_string = input("Enter a string: ")

no_punctuations = input_string.translate(
    str.maketrans("", "", string.punctuation)
)

print("String without punctuations:", no_punctuations)
```
