```python
# EXPERIMENT 23
# Alpha-Beta Pruning Algorithm

def alpha_beta(depth, node, maximizing, values, alpha, beta):


    if depth == 3:
        return values[node]

    if maximizing:

        best = float('-inf')

        for child in [node * 2, node * 2 + 1]:

            value = alpha_beta(
                depth + 1,
                child,
                False,
                values,
                alpha,
                beta
            )

            best = max(best, value)
            alpha = max(alpha, best)


            if beta <= alpha:
                break

        return best

    else:

        best = float('inf')

        for child in [node * 2, node * 2 + 1]:

            value = alpha_beta(
                depth + 1,
                child,
                True,
                values,
                alpha,
                beta
            )

            best = min(best, value)
            beta = min(beta, best)


            if beta <= alpha:
                break

        return best



values = [
    3, 5,
    2, 9,
    12, 5,
    23, 23
]

result = alpha_beta(
    0,
    0,
    True,
    values,
    float('-inf'),
    float('inf')
)

print("Optimal value using Alpha-Beta Pruning:", result)
```
