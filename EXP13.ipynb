```python
# EXPERIMENT 13
# Crypt-Arithmetic Problem

from itertools import permutations


def solve_cryptarithmetic():
    letters = "SENDMORY"

    for perm in permutations(range(10), len(letters)):

        mapping = dict(zip(letters, perm))


        if mapping['S'] == 0 or mapping['M'] == 0:
            continue

        SEND = (
            mapping['S'] * 1000 +
            mapping['E'] * 100 +
            mapping['N'] * 10 +
            mapping['D']
        )

        MORE = (
            mapping['M'] * 1000 +
            mapping['O'] * 100 +
            mapping['R'] * 10 +
            mapping['E']
        )

        MONEY = (
            mapping['M'] * 10000 +
            mapping['O'] * 1000 +
            mapping['N'] * 100 +
            mapping['E'] * 10 +
            mapping['Y']
        )

        if SEND + MORE == MONEY:
            return mapping

    return None


solution = solve_cryptarithmetic()

if solution:
    print("Solution:")
    print(solution)

    print("\nSEND =",
          solution['S']*1000 +
          solution['E']*100 +
          solution['N']*10 +
          solution['D'])

    print("MORE =",
          solution['M']*1000 +
          solution['O']*100 +
          solution['R']*10 +
          solution['E'])

    print("MONEY =",
          solution['M']*10000 +
          solution['O']*1000 +
          solution['N']*100 +
          solution['E']*10 +
          solution['Y'])
else:
    print("No solution found.")
```
