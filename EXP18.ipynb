```python
# EXPERIMENT 18
# Travelling Salesman Problem

import numpy as np

def tsp(cities):

    n = len(cities)


    dist_matrix = np.zeros((n, n))

    for i in range(n):
        for j in range(n):

            if i != j:
                dist_matrix[i][j] = np.linalg.norm(
                    cities[i] - cities[j]
                )



    tour = [0]

    unvisited = set(range(1, n))

    total_distance = 0


    while unvisited:

        current_city = tour[-1]

        nearest_city = min(
            unvisited,
            key=lambda city: dist_matrix[current_city][city]
        )

        tour.append(nearest_city)

        unvisited.remove(nearest_city)

        total_distance += dist_matrix[
            current_city
        ][nearest_city]


    tour.append(0)

    total_distance += dist_matrix[
        tour[-2]
    ][0]

    return tour, total_distance



cities = np.array([
    [0, 0],
    [2, 3],
    [5, 2],
    [6, 6],
    [1, 5]
])


tour, distance = tsp(cities)

print("Best Tour:", tour)
print("Total Distance:", round(distance, 2))
```
