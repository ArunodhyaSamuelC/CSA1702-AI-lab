```python
# EXPERIMENT 15
# Vacuum Cleaner Problem

def vacuum_cleaner():

    rooms = {
        "A": "Dirty",
        "B": "Dirty"
    }

    current_room = "A"

    print("Initial State:")
    print(rooms)
    print("Vacuum is in room:", current_room)
    print()


    print("Vacuum cleans Room A")
    rooms["A"] = "Clean"


    print("Vacuum moves from Room A to Room B")
    current_room = "B"


    print("Vacuum cleans Room B")
    rooms["B"] = "Clean"

    print("\nFinal State:")
    print(rooms)
    print("Vacuum is in room:", current_room)


vacuum_cleaner()
```
