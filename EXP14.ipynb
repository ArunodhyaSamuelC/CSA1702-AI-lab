```python
# EXPERIMENT 14
# Missionaries and Cannibals Problem

from collections import deque


def is_valid(m_left, c_left, m_right, c_right):


    if min(m_left, c_left, m_right, c_right) < 0:
        return False


    if m_left > 0 and m_left < c_left:
        return False


    if m_right > 0 and m_right < c_right:
        return False

    return True


def solve():



    start = (3, 3, 0)
    goal = (0, 0, 1)

    queue = deque()
    queue.append((start, []))

    visited = {start}


    moves = [
        (1, 0),
        (2, 0),
        (0, 1),
        (0, 2),
        (1, 1)
    ]

    while queue:

        state, path = queue.popleft()

        ml, cl, boat = state

        if state == goal:
            return path

        for m, c in moves:

            if boat == 0:

                new_ml = ml - m
                new_cl = cl - c
                new_boat = 1

            else:

                new_ml = ml + m
                new_cl = cl + c
                new_boat = 0

            new_mr = 3 - new_ml
            new_cr = 3 - new_cl

            if is_valid(new_ml, new_cl, new_mr, new_cr):

                new_state = (
                    new_ml,
                    new_cl,
                    new_boat
                )

                if new_state not in visited:

                    visited.add(new_state)

                    action = (
                        f"{m} Missionary(s), "
                        f"{c} Cannibal(s)"
                    )

                    queue.append((
                        new_state,
                        path + [action]
                    ))

    return None


solution = solve()

if solution:

    print("Solution found!\n")

    for i, step in enumerate(solution, 1):
        print(i, ".", step)

else:
    print("No solution found.")
```
