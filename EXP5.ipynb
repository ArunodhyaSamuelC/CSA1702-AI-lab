```python
# EXPERIMENT 5
# List Operations

# Nested List
nested_list = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print("Nested List:", nested_list)

# Length
my_list = [1, 2, 3, 4, 5]
print("Length:", len(my_list))

# Concatenation
list1 = [1, 2, 3]
list2 = [4, 5, 6]

concatenated = list1 + list2
print("Concatenated List:", concatenated)

# Membership
if 3 in my_list:
    print("3 is in the list")
else:
    print("3 is not in the list")

# Iteration
print("Iteration:")
for item in my_list:
    print(item)

# Indexing
print("First element:", my_list[0])
print("Last element:", my_list[-1])

# Slicing
print("Sliced List:", my_list[1:4])
```
